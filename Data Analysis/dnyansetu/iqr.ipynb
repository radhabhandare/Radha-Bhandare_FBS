```python
import numpy as np
data = [1,2,4,5,6,89]
q1 = np.percentile(data,25)
q3 = np.percentile(data,75)
iqr = q3-q1
lower = q1- 1.5*iqr
upper = q3+ 1.5*iqr

print("lower bound", lower)
print("upper bound", upper)

outlier = []
for i in data:
    if i<lower or i>upper:
        outlier.append(i)
print("outlier", outlier)
```
