```python
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = pd.read_csv("C:/Users/Radha/OneDrive/Desktop/FBS/Datasets-20261026T112107Z-1-001/Datasets/Iris.csv")
data.head()
```

### Facetgrid 
1. same graph will displayed with categories
2. create multiple grapgh based onn subset of data


```python
g = sns.FacetGrid(data, col = 'Species')
g.map(sns.scatterplot, 'Petal_Length', 'Petal_Width')
plt.show()
```

###Pairplot
1. Relationship between numerical column
2. Gives distribution for diagonal
3. Scatterplot for relationship between other column
   

```python
sns.pairplot(data, hue ='Species')
```
