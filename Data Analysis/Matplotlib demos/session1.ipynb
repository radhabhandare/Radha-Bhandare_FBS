## Lineplot
1. Time series analysis
2. trends over time
3. upword and downword trends
4. example IPl team scoresheet


```python
overs = np.arange(1,21)
team1_runs = np.random.randint(0,37,20)
team2_runs = np.random.randint(0,37,20)
```

```python
team1_score = np.cumsum(team1_runs)
team2_score = np.cumsum(team2_runs)
```

```python
team1_runs
```

```python
team1_score
```

```python
plt.style.use('dark_background')
plt.plot(overs, team1_score, color = 'red', label = 'Rcb')
plt.plot(overs, team2_score, color = 'yellow', label = 'Csk')
plt.xlabel('Overs')
plt.ylabel('Runs')
plt.title('Scoresheet')
plt.legend()
plt.show()
```
